#Set-up Project Environment for GizmoBox Data Lakehouse

###1. Access the container gizmobox

```sql
%fs ls 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/landing'
```

###2. Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS gizmobox
  URL 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/'
  WITH (STORAGE CREDENTIAL dea_termi_ext_sc)
  COMMENT 'External Location for Gizmobox'
```

###3. Create the catalog - gizmobox

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS gizmobox_ct
    MANAGED LOCATION 'abfss://termi-001@databrickstermiextdl.dfs.core.windows.net/'
    COMMENT 'This is a catalog for Gizmobox Data Lakehouse'
```

###4. Create Schemas
1. Landing
2. Bronze
3. Silver
4. Gold

```sql
SELECT current_catalog()
```

```sql
USE CATALOG gizmobox_ct;
```

```sql
SELECT current_catalog();
```

```sql
CREATE SCHEMA IF NOT EXISTS landing
    COMMENT 'This is a schema for landing zone'
    MANAGED LOCATION 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/landing'
```

```sql
SHOW SCHEMAS;
```

```sql
CREATE SCHEMA IF NOT EXISTS bronze
    COMMENT 'This is a schema for bronze zone'
    MANAGED LOCATION 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS silver
    COMMENT 'This is a schema for silver zone'
    MANAGED LOCATION 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS gold
    COMMENT 'This is a schema for gold zone'
    MANAGED LOCATION 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/gold';
```

```sql
SHOW SCHEMAS;
```

```sql
USE CATALOG gizmobox_ct;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
  LOCATION 'abfss://gizmobox@databrickstermiextdl.dfs.core.windows.net/landing/operational-data'
```
